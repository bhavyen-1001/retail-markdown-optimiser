# Elasticity Modelling and Validation

The purpose of this notebook is to estimate price elasticity for each product segment.<br>
We fit two Ordinary Least Squares (OLS) models per segment: 
<ul><li>A baseline model (price only)
<li>A controlled model (price + seasonality controls).</ul>
By comparing the two, we demonstrate that controlling for seasonality (week number and Q4 spikes) is essential to avoid confounding and biased elasticity estimates. We then run statistical diagnostics and walk-forward cross-validation on the controlled model to ensure the estimates are robust and predictive before passing them to the optimisation engine.
